### Exercício 1.1: Shapes e Reshape (O erro mais comum no Scikit-Learn)
Crie um array 1D com os números de 1 a 12. Em seguida:
1. Exiba sua dimensão (`shape`).
2. Transforme-o em uma matriz $4 \times 3$.
3. Transforme-o em um vetor coluna de formato $(12, 1)$ usando `-1`.
*Por que o Scikit-Learn exige uma matriz 2D $(N, 1)$ para a matriz de features $X$, mas aceita um vetor 1D $(N,)$ para o vetor alvo $y$?*

```python
import numpy as np

# 1. Criar array de 1 a 12
arr = np.arange(1, 13)
print("Shape original:", arr.shape)

# 2. Reshape para (4, 3)
matriz_4x3 = arr.reshape(4, 3)
print("\nMatriz 4x3:\n", matriz_4x3)

# 3. Reshape para coluna (12, 1)
vetor_coluna = arr.reshape(-1, 1)
print("\nVetor coluna shape:", vetor_coluna.shape)
```

### Exercício 1.2: Filtragem Booleana (Masking)
Dado um array de predições de um modelo contínuo e os valores reais, filtre apenas as amostras onde o erro absoluto for maior que 2.0.

```python
y_true = np.array([10.5, 14.0, 19.2, 22.1, 8.0])
y_pred = np.array([10.0, 17.5, 18.9, 19.0, 11.2])

# Calcule o erro absoluto para cada amostra
errors = np.abs(y_true - y_pred)

# Filtre apenas onde o erro absoluto > 2.0 usando máscara booleana
mask = errors > 2.0
print("Máscara booleana:", mask)
print("Amostras reais com grande erro:", y_true[mask])
print("Erros encontrados:", errors[mask])
```

### Exercício 1.3: Implementando Métricas na Mão (MSE, RMSE, MAE)
No Projeto 1 da usina, você precisará analisar $R^2$, RMSE e MAE.
Antes de chamar `sklearn.metrics`, vamos calcular na mão com operações vetorizadas do NumPy para garantir que você entende a fórmula exata:
- $MAE = \frac{1}{N} \sum_{i=1}^N |y_i - \hat{y}_i|$
- $MSE = \frac{1}{N} \sum_{i=1}^N (y_i - \hat{y}_i)^2$
- $RMSE = \sqrt{MSE}$

```python
mae = np.mean(np.abs(y_true - y_pred))
mse = np.mean((y_true - y_pred) ** 2)
rmse = np.sqrt(mse)

print(f"MAE:  {mae:.4f}")
print(f"MSE:  {mse:.4f}")
print(f"RMSE: {rmse:.4f}")
```
