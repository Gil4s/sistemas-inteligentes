### Questão 1: 

#### Você recebeu os dados cadastrais de 6 solicitantes de crédito. O conjunto contém variáveis numéricas, variáveis categóricas (texto), um registro com valor ausente e a variável alvo binária indicando se o empréstimo foi aprovado ou recusado.

```python
import numpy as np
import pandas as pd

dados_emprestimo = {
    'idade': [23, 45, 31, 52, 28, 40],
    'renda_mensal': [2500.0, 7200.0, 3100.0, np.nan, 2100.0, 5400.0],
    'finalidade': ['carro', 'imovel', 'educacao', 'imovel', 'carro', 'educacao'],
    'possui_garantia': ['sim', 'nao', 'nao', 'sim', 'nao', 'sim'],
    'aprovado': [0, 1, 0, 1, 0, 1]
}
```

```python
## Tarefa 1 -> Criar um DataFrame df_emprestimo a partir do dicionário

## método simples do pandas para a criação do dataframe que funciona bem em dicionários
df = pd.DataFrame(dados_emprestimo)
print(df)

## o isna() é um método que valida os valores nulos de alguma coisa.
## a ideia é validar se são nulos ou não ... 
## se true, eu uso o metodo de soma para somar esses valores nulos e já entrego
print ("")
idade_colum = (df['idade']).isna().sum()
print (idade_colum) 

print ("")
renda_colum = (df['renda_mensal']).isna().sum()
print (renda_colum) 

print ("")
finalidade_colum = (df['finalidade']).isna().sum()
print (finalidade_colum) 

print ("")
garantia_colum = (df['possui_garantia']).isna().sum()
print (garantia_colum) 

print ("")
aprovado_colum = (df['aprovado']).isna().sum()
print (aprovado_colum) 


## o problema é que dessa forma, tudo é expresso em coluna
```

```python

idade_colum = (df['idade']).isna().sum()
print (idade_colum) 

print ("")
renda_colum = (df['renda_mensal']).isna().sum()
print (renda_colum) 

print ("")
finalidade_colum = (df['finalidade']).isna().sum()
print (finalidade_colum) 

print ("")
garantia_colum = (df['possui_garantia']).isna().sum()
print (garantia_colum) 

print ("")
aprovado_colum = (df['aprovado']).isna().sum()
print (aprovado_colum) 
```

```python
## o que da para fazer, é validar pelo dataframe inteiro, isso o pandas permite, sem que eu acesse direto na coluna

print(df.isna().sum())

## mais fácil e fica mais organizado
```

```python
## Tarefa 2 -> Vamos remover as linhas com valores ausentes


print (df)
print (f"\n{df.shape}\n") ## só praticando mesmo ^^ 

## Aqui, usando o df.dropna(), eu estou limpando as tuplas que possuem valores NULL ...
## É uma forma de "limpar" o meu conjunto de dados caso eu interprete que isso não é tão relevante assim
df_limpo = df.dropna()
print (df_limpo)

## Vamos apresentar os resultados das quantidade de linhas e colunas novas que temos em nosso dataframe
print ("")
print (df_limpo.shape)

## Resultamos na eliminação de uma tupla ... 

```

```python
## Tarefa 3 -> Separar atributos alvo (aprovado limpo e apresentar ele separadamente)

x = df.drop(columns = ['aprovado'])
print (x)


## Apresentando a coluna de aprovado separadamente
y = df['aprovado']
print (f"\n{y}") ## só para praticar o fprint ^^
```

```python
## Tarefa 4 -> codificar duas colunas (finalidade e possui_garantia, definindo dtype = int)

## Uma forma de fazer esse trabalho com das colunas inteiras, é concatenando elas com o operador de soma (+) dentro do get_dummies
x_tratado = pd.get_dummies(x, columns = ['finalidade'] + ['possui_garantia'], dtype = int)
print (x_tratado)
```

```python
## Tarefa 5 -> A partir de df_limpo, vamos filtrar e exibit apenas os solicitante que possuem renda_mensal > 3000.0

## Para isso, precisamos inicialmente criar uma mascara
filtro_df = df_limpo['renda_mensal'] > 3000.0 

## apresentação da máscara
print (filtro_df)

## Voltar com as tuplas necessárias

resultado = df_limpo [filtro_df]
print (f"\n{resultado}")
```
