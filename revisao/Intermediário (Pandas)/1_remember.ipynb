## 3. O que é One-Hot Encoding (`pd.get_dummies`)?

### O Problema: Modelos de IA não sabem fazer contas com palavras
Se você olhar para a nossa matriz `x` criada na célula anterior, temos a coluna `Historico` com textos: `'bom'` e `'ruim'`.
- Modelos de Machine Learning calculam distâncias, produtos escalares e derivadas.
- Você **não pode** multiplicar `peso * 'bom'`.

### Por que NÃO trocar por 1 e 2?
Se trocássemos `'bom' -> 1` e `'ruim' -> 2`, o modelo matemático acharia que `'ruim'` vale **o dobro** de `'bom'`, ou que existe uma ordem numérica entre eles. Como são categorias puras (nominais), isso distorceria a matemática do algoritmo.

---

### A Solução: A Analogia com Circuitos Digitais (One-Hot)
> Em **Circuitos Digitais**, um barramento **One-Hot** é aquele onde **apenas 1 bit fica em nível alto (`1`) e todos os outros ficam em nível baixo (`0`)**.

O `pd.get_dummies()` faz exatamente isso: ele **remove a coluna de texto** e cria **uma nova coluna para cada categoria**, preenchendo apenas com `0` e `1` (como interruptores independentes).

#### Tabela Antes vs Depois do One-Hot Encoding:

**Antes (com texto):**

| Idade | Saldo | Historico |
| :---: | :---: | :---: |
| 25 | 1200.0 | **bom** |
| 45 | 3500.0 | **ruim** |
| 30 | 500.0 | **bom** |
| 50 | 8000.0 | **bom** |

**Depois com `pd.get_dummies()`:**

| Idade | Saldo | `Historico_bom` | `Historico_ruim` |
| :---: | :---: | :---: | :---: |
| 25 | 1200.0 | **1** | **0** |
| 45 | 3500.0 | **0** | **1** |
| 30 | 500.0 | **1** | **0** |
| 50 | 8000.0 | **1** | **0** |

> **Aplicação Direta no Projeto 4 (German Credit):** O dataset de crédito tem 13 colunas de texto (finalidade do empréstimo, tipo de emprego, moradia, etc.). Usamos o `pd.get_dummies()` para transformar todas essas palavras em matrizes numéricas binárias antes de treinar a Árvore de Decisão!

```python
## Aplicando o One-Hot Encoding na pratica sobre a matriz x:
## O parametro columns=['Historico'] diz qual coluna queremos transformar em binario
## O parametro dtype=int garante que os valores serao 0 e 1 inteiros (em vez de True/False)

x_codificado = pd.get_dummies(x, columns=['Historico'], dtype=int)
## O interessante é que, usando o método, a gente já cria as colunas baseadas nos dados catagóricos
## exemplo -> Coluna é Histórico. 
## é categorizado como bom ou ruim
## as novas colunas são: Histórico + "_" + categoria
## Logo: Histórico_bom | Histórico_ruim

print('--- MATRIZ X APOS O ONE-HOT ENCODING ---')
display(x_codificado)
```

### Detalhe Técnico: O Parâmetro `drop_first=True`
Repare que se o cliente **não tem** histórico `'ruim'` (`0`), ele **obrigatoriamente** tem histórico `'bom'` (`1`). Ter as duas colunas é uma informação redundante (em estatística, isso pode gerar multicolinearidade).

Para evitar colunas redundantes, usamos frequentemente o parâmetro `drop_first=True`, que remove a primeira categoria e deixa apenas $K - 1$ colunas:

```python
## Com drop_first=True, eliminamos a coluna redundante:
x_otimizado = pd.get_dummies(x, columns=['Historico'], drop_first=True, dtype=int)

print('--- MATRIZ X COM drop_first=True ---')
display(x_otimizado)
```
