```python
## Vale salientar (somente para lembrar mesmo e revisar escrevendo)
## Um dos motivos pelo qual usamos os vetores em python e não uma lista para operar (por mais que pareça óbvio) mas é justamente pelo fato de termos como trabalhar diretamente com os valroes usando artifícios de algumas bibliotecas, além de ser um componente muito usado em álgebra lienar
## Conseguimos trabalhar com matrizes por meio de fetores e operar também com base nelas
## Durante o estudo, artifícios como esses, são extremamente valiosos

## agora, vamos fazer a miha matriz, com mais de uma linha

matriz = np.array([
    [10, 20], 
    [30, 40]
]) ## Veja que, para cada linha, eu tive que criar um novo colchete, ou seja, para cada vetor, eu preciso fazer isso, para que de fato seja interpertado como um novo vetor ali dentro

print (matriz)

## Imagine que eu tenho um modelo com várias características (representadas por atributos pro exemplo) 
## Cada característica, eu represento como uma coluna (referência a banco de dados)
## Então eu precisaria configurar corretamente as colunas por meio de uma matriz (caso eu quisesse) para trabalhar com essas características corretamente
## Logo, a ideia era eu ter minhas tuplas (as entradas ou os dados por si só por exemplo) tento na linha, suas respectivas características valoradas (ou não) por exemplo ... 
## Mas e se eu tenho somente uma característica?
## Minha matriz seria uma MATRIZ COLUNA, sendo representada somente por uma coluna
```

```python
## Significado de um ".reshape(-1, 1)"

## A função .reshape(linhas, colunas) reorganiza a geometria do array. Quando você passa -1, você está dizendo ao NumPy:

# "Numpy, calcule você mesmo a quantidade de linhas baseada no tamanho total dos dados, e me garanta que o resultado terá exatamente 1 coluna!"

## Transformando o nosso vetor (o que eu criei na primeira janela do notebook) em um vetor mono coluna (antes ele era uma tupla com 4 colunas)

coluna = vetor.reshape(-1, 1)

## Como funciona: Se eu tenho um array com 4 valores por exemplo, o reshape pode fazer algumas operações

## 1 x 4 = 4 -> .reshape (1, 4)
## 2 X 2 = 4 -> .reshape (2, 2)

## o -1 é um coringa (é uma icógnica X da matemática) 

## Basicamente eu digo ao NumPy para calcular o valor para mim
## .reshape(-1, 1)
## Linhas = (total de elementos / colunas)

## agora vamos para a prática

## Se eu cravei que o meu elemento pode ter somente uma coluna, por lógica, o calculo da minha quantidade de linhas vai acontecer com base no novo valor de colunas que eu disse que queria

## Então se eu der um reshape, colocar o "-1" como parâmetro, no calculo, a nova quantidade de colunas já vai ser incluida, então basicamente eu to convertendo tudo para ficar em monocoluna
```
