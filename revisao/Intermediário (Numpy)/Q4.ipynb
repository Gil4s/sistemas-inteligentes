```python
import numpy as np

## criando uma mascara booleana 

## Quando aplicamos um operador >, <, ==, !=, diretamente sobre um array, o Numpy não compara o array inteir com um bloco. Ele compara elemento por elemento em C e devolve um novo array contendo apenas True e False
temperaturas = np.array([22.0, 31.5, 19.8, 33.2, 25.0])

mascara = temperaturas > 30

print (mascara)

## Veja que, se eu rodar isso aquim eu basicamente vou armazenar o que eu tenho em booleano. 

## Então se o valor é ou não maior, eu tenho um retorno.


```

```python
## Indexação Booleana (filtrando os dados)

## O array original filtrado pela mascara:

print (mascara)

temperaturas_altas = temperaturas [mascara]

print (temperaturas_altas)

## Nesse caso, eu estou armazenando dentro da variável, o novo array. 
## veja que, temperaturas é um array, mas eu estou passando como parâmetro a "mascara" que eu representei com um print usando o operador lógico
## A saída não é em booleano justamente pelo fato de não armazenar expressando em booleano


## O que acotnece aqui (mascara): 

## 1 -> Veja que eu armazenei em máscara, usando uma condição em booleano
## 1.2 -> A minha saída vai ser interpretada em booleano 
## 1.3 -> Mas ainda sim, eu tenho um vetor representado pela aquela condição, pois eu operei em ciam de um vetor e a minha saída é expressada em cima de um vetor resultante da operação
## 1.4 -> Embora a minha saída seja em boolean, temos um vetor criado a partir da condição, expresso na estrutura de boolean

## E o que acontece com "temperaturas_altas"? 

## 2 -> como eu estou passando que, temperaturas altas é um vetor pois estou igualando a temperaturas (outro vetor que temos)
## 2.1 -> Isso retrata que o valor armazenado vai ser um vetor. Logo, o vetor vai ser o mesmo de temperaturas, com as condições de mascara
## 2.2 -> Resultado é expresso com um vetor, somente com os valores TRUE dado por máscara, mas não em booleano, pois eu não estou operando, apenas armazenando valores que passam por condições
```

```python
## Computacionalmente, os valores em booleano são como 1 e 0

## Então se eu somar um array booleano com o "np.sum()", o NumPy soma os 1, dando exatamente a quantidade de elementos que atendem a condição
## Tipo, como os "true" são 1, todo "true" que passar da condição, é somado, então é uma forma simples por exemplo, de aplicar um filtro .... 
## Isso de certa forma, pode ajudar no pre-processamente e é uma ideia interessante na parte de implementação trabalhar dessa forma

qtd_critica = np.sum(mascara) ## Conta: 0 + 1 + 0 + 1 + 0 = 2
print (f"Total de snsores a cima de 30ºC : {qtd_critica}")
```

```python
## O if Ternário Vetorizado: n´.where(condição, valor_se_V, valor_se_f)

## Se temperatura > 30, atribui 1 (alerta); senão, atribui 0 (normal):
status_binario = np.where(temperaturas > 30.0, 1, 0)
print (status_binario)

## Então basicamente, o np where faz como que uma arvore mesmo

## é interessante, justamente pelo fato da simplicidade 
## então se eu quero criar condições no ssenido de cada altura da minha árvore, trabalhar com o np.where, além de ser bem funcional, é interessante para e fácil para criar as camadas

# Saida: [0, 1, 0, 1, 0] 

## Basicamente, essa é a lógica que usamos para trabalhar a questão do problema 2, qpara quando eu tenho 0 para paciêntes saudáveis e 1 para paciêtes doentes
```

```python
## Questão 4:  Doenças Cardíacas - Heart Disease 
## no problema 2 do projeto, temosu ma variável clinica importante, que é o nível do colesterol sérico (chol). Níveis de colesterol acima de 200.0 mg/dl são considerados clinicamente elevados e colocam o paciente em um grupo de atenção


import numpy as np

colesterol = np.array ([195.0, 240.0, 210.0, 180.0, 260.0, 290.0, 175.0, 230.0])

## vamos criar uma mascara booleana, basicamente retornando valor em boolean com base nas condições que queremos para poder trabalhar isso depois

mascara_risco = colesterol > 200
print (mascara_risco) 

## Agora, vamos filtrar os valores usando uma indexação booleana e vamos imprimir de fato os valores que atendem isso, em um vetor novo

filtro_risco = colesterol[mascara_risco]
print (filtro_risco)

## A agora, vamos contar os casos de risco.
## Vamos calcular a quantidade de paciêntes em risco nesse caso

pacientes_risco = np.sum(mascara_risco) 
print (pacientes_risco)

binario_risco = np.where(colesterol > 200, 1, 0)
print (binario_risco)


```
