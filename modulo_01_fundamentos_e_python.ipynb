## 1. Revisão do Básico do Python: Listas, Dicionários e List Comprehension
No Aprendizado de Máquina, frequentemente manipulamos coleções de dados. Veja a forma tradicional vs a forma pythônica:

```python
# 1.1 Listas e List Comprehension
temperaturas_celsius = [20.5, 25.0, 31.2, 18.4, 27.8]

# Forma tradicional (loop for)
temp_kelvin_loop = []
for t in temperaturas_celsius:
    temp_kelvin_loop.append(t + 273.15)

# Forma Pythônica (List Comprehension - mais rápida e limpa!)
temp_kelvin = [t + 273.15 for t in temperaturas_celsius]

print('Temperaturas em Celsius:', temperaturas_celsius)
print('Temperaturas em Kelvin: ', temp_kelvin)
```

## 2. NumPy: A Base Numérica do Machine Learning
Em IA, raramente usamos listas nativas do Python para cálculos matemáticos porque listas são lentas e consomem muita memória.
Usamos o **NumPy** (
p.ndarray), que executa cálculos em C e permite **vetorização** (aplicar operações em matrizes inteiras sem nenhum loop or).

```python
import numpy as np

# Criando uma matriz de dados X (4 amostras, 2 atributos: Horas de Estudo e Nota no Simulado)
# Cada linha = 1 aluno (amostra/instância)
# Cada coluna = 1 atributo (feature)
X = np.array([
    [2.0, 45.0],
    [5.0, 70.0],
    [8.0, 85.0],
    [1.5, 30.0]
])

print('Formato da matriz (linhas, colunas):', X.shape)
print('Primeira linha (primeiro aluno):', X[0, :])
print('Primeira coluna (Horas de estudo de todos):', X[:, 0])

# Operação Vetorizada: Calculando a média de cada coluna instantaneamente
medias = np.mean(X, axis=0)
desvios = np.std(X, axis=0)

print('\nMédia de cada atributo:', medias)
print('Desvio padrão de cada atributo:', desvios)
```

## 3. Pandas: Tabelas Estruturadas (DataFrames)
O **Pandas** é a ferramenta que organiza dados em tabelas com nomes de colunas, exatamente como uma planilha do Excel ou tabela SQL.
É a biblioteca padrão que usaremos para carregar os datasets dos 5 projetos (.csv, .data, etc.).

```python
import pandas as pd

# Criando um DataFrame de exemplo simulando um conjunto de dados médico
dados = {
    'idade': [25, 45, np.nan, 35, 60],
    'pressao_arterial': [120, 140, 130, np.nan, 150],
    'fumante': ['Nao', 'Sim', 'Nao', 'Sim', 'Sim'],
    'doenca_cardiaca': [0, 1, 0, 0, 1]  # Nosso alvo (target) y
}

df = pd.DataFrame(dados)
print('--- Visualizando o DataFrame ---')
display(df)
```

## 4. Pré-Processamento: Tratamento de Valores Ausentes (NaN)
Como vimos nos slides:
- Para atributos numéricos: usamos a **média** ou a **mediana** (a mediana é mais robusta a outliers).
- Para atributos categóricos: usamos a **moda** (o valor mais frequente).
- Se entra lixo, sai lixo (*garbage in, garbage out*).

```python
# Verificando quantos valores ausentes existem em cada coluna
print('Valores nulos por coluna:\n', df.isnull().sum())

# Imputação pela mediana na coluna 'idade'
mediana_idade = df['idade'].median()
df['idade'] = df['idade'].fillna(mediana_idade)

# Imputação pela média na coluna 'pressao_arterial'
media_pressao = df['pressao_arterial'].mean()
df['pressao_arterial'] = df['pressao_arterial'].fillna(media_pressao)

print('\n--- DataFrame Limpo ---')
display(df)
```
